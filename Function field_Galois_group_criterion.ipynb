```python
# Polynomial ring
R.<x> = PolynomialRing(QQ)

# Fibonacci polynomial F_n(x)
def fibonacci_polynomial(n):
    if n == 0:
        return R(0)
    if n == 1:
        return R(1)

    F0 = R(0)
    F1 = R(1)

    for k in range(2, n + 1):
        F0, F1 = F1, x*F1 + F0

    return F1


# Compute F_{p+1}, its first derivative, the roots of the derivative,
# and the values of F_{p+1} at those roots
def critical_values_fibonacci(p, root_ring=QQbar):
    F = fibonacci_polynomial(p + 1)
    dF = F.derivative()

    # roots of the first derivative
    roots_dF = dF.roots(ring=root_ring, multiplicities=False)

    # values of F_{p+1} at the roots of F'_{p+1}
    values = [(r, F(r)) for r in roots_dF]

    return F, dF, roots_dF, values


# Example
p = 3

F, dF, roots_dF, values = critical_values_fibonacci(p)

print("F_{p+1}(x) =", F)
print("F'_{p+1}(x) =", dF)
print("Roots of F'_{p+1}(x) =", roots_dF)
print("Values of F_{p+1} at the roots of F'_{p+1}:")
for r, v in values:
    print("root =", r, "   F_{p+1}(root) =", v)
```
